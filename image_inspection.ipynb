```python
zip_path = '/content/drive/MyDrive/Independent Study/3. Brain Tumor Project/data/raw/1512427.zip'
```

```python
import zipfile
import h5py
import io
import numpy as np
import matplotlib.pyplot as plt

#Opening original dataset
with zipfile.ZipFile(zip_path, 'r') as archive:

    # Getting the first subset
    nested_bytes = archive.read('brainTumorDataPublic_1-766.zip')

# Opening the subset
with zipfile.ZipFile(io.BytesIO(nested_bytes), 'r') as inner:

    # Reading first MRI file
    mat_bytes = inner.read('1.mat')

#Loading MATLAB data
with h5py.File(io.BytesIO(mat_bytes), 'r') as f:

    cjdata = f['cjdata']

    image = cjdata['image'][:]
    label = int(cjdata['label'][0, 0])
    pid_raw = cjdata['PID'][:]
    tumor_mask = cjdata['tumorMask'][:]

print("Image shape:", image.shape)
print("Label:", label)
print("Raw patient ID:", pid_raw.flatten())
print("Tumor mask shape:", tumor_mask.shape)
```

```python
tumor_types = {
    1: "Meningioma",
    2: "Glioma",
    3: "Pituitary tumor"
}

print("Tumor type:", tumor_types[label])
```
